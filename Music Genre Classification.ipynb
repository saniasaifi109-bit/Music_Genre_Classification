## 7 - Data Preprocessing

```python
# The supplied feature dataset already contains extracted numerical audio features.
# 'filename' identifies the audio segment and 'label' is the target genre.
print("Duplicate rows:", data.duplicated().sum())
print("Genre classes:", sorted(data['label'].unique()))
print("Number of genres:", data['label'].nunique())
```

### A - Outlier Inspection

```python
import matplotlib.pyplot as plt
import seaborn as sns

sns.boxplot(x=data['tempo'])
plt.title('Tempo Outlier Inspection')
plt.show()
```

Audio features naturally have different ranges and musical variation.
Therefore, no aggressive row deletion is performed only because a value appears extreme.
StandardScaler is used with Logistic Regression, SVM and KNN to handle scale differences.

### B - Encoding the Target Column

```python
from sklearn.preprocessing import LabelEncoder

label_encoder = LabelEncoder()
data['label_encoded'] = label_encoder.fit_transform(data['label'])

print("Genre classes:", list(label_encoder.classes_))
```

## 8 - Store Feature Matrix In X and Response(Target) In Vector y

```python



# Use every numerical audio feature supplied by the current feature dataset.
# filename and label are not model features.
feature_columns = [
    c for c in data.columns
    if c not in ['filename', 'label', 'label_encoded']
]

X = data[feature_columns]
y = data['label_encoded']

print("Number of features:", len(feature_columns))
print("Feature shape:", X.shape)
print("Target shape:", y.shape)
print("Features:", feature_columns)
```

## 9 - Splitting The Dataset Into The Training Set And Test Set

```python
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.20, random_state=42, stratify=y
)

print("Training rows:", len(X_train))
print("Testing rows:", len(X_test))
```

## 10 - Import The models

```python
from sklearn.linear_model import LogisticRegression
from sklearn.svm import SVC
from sklearn.tree import DecisionTreeClassifier
from sklearn.neighbors import KNeighborsClassifier
from sklearn.ensemble import RandomForestClassifier
from sklearn.pipeline import Pipeline
from sklearn.preprocessing import StandardScaler
```

## 11 - Model Training

```python
lr = Pipeline([
    ('scaler', StandardScaler()),
    ('model', LogisticRegression(C=0.5, max_iter=5000))
])
lr.fit(X_train, y_train)

svm = Pipeline([
    ('scaler', StandardScaler()),
    ('model', SVC(C=5, probability=True, random_state=42))
])
svm.fit(X_train, y_train)

dt = DecisionTreeClassifier(random_state=42)
dt.fit(X_train, y_train)

knn = Pipeline([
    ('scaler', StandardScaler()),
    ('model', KNeighborsClassifier(n_neighbors=5))
])
knn.fit(X_train, y_train)

rf = RandomForestClassifier(
    n_estimators=300,
    random_state=42,
    n_jobs=-1
)
rf.fit(X_train, y_train)

print("All five models trained successfully.")
```

## 12 - Prediction on Test Data

```python
y_pred1 = lr.predict(X_test)
y_pred2 = svm.predict(X_test)
y_pred3 = dt.predict(X_test)
y_pred4 = knn.predict(X_test)
y_pred5 = rf.predict(X_test)
```

## 13 - Evaluating the Algorithm

```python
from sklearn import metrics

score1 = metrics.accuracy_score(y_test, y_pred1)
score2 = metrics.accuracy_score(y_test, y_pred2)
score3 = metrics.accuracy_score(y_test, y_pred3)
score4 = metrics.accuracy_score(y_test, y_pred4)
score5 = metrics.accuracy_score(y_test, y_pred5)

print("Logistic Regression Accuracy:", score1)
print("SVM Accuracy:", score2)
print("Decision Tree Accuracy:", score3)
print("KNN Accuracy:", score4)
print("Random Forest Accuracy:", score5)
```

```python
final_data = pd.DataFrame({
    'Models': [
        'Logistic Regression',
        'SVM',
        'Decision Tree',
        'K-Nearest Neighbour',
        'Random Forest'
    ],
    'Accuracy': [score1, score2, score3, score4, score5]
})

final_data.sort_values('Accuracy', ascending=False)
```

```python
sns.barplot(x=final_data['Models'], y=final_data['Accuracy'])
plt.xticks(rotation=20)
plt.ylim(0, 1)
plt.title('Model Accuracy Comparison')
plt.ylabel('Accuracy')
plt.show()
```

```python
from sklearn.metrics import accuracy_score, precision_score, recall_score, f1_score, confusion_matrix
```

```python
# Detailed metrics for all five models
metric_rows = []

predictions = {
    'Logistic Regression': y_pred1,
    'SVM': y_pred2,
    'Decision Tree': y_pred3,
    'K-Nearest Neighbour': y_pred4,
    'Random Forest': y_pred5
}

for name, pred in predictions.items():
    metric_rows.append({
        'Model': name,
        'Accuracy': accuracy_score(y_test, pred),
        'Precision': precision_score(y_test, pred, average='weighted', zero_division=0),
        'Recall': recall_score(y_test, pred, average='weighted', zero_division=0),
        'F1 Score': f1_score(y_test, pred, average='weighted', zero_division=0)
    })

metrics_data = pd.DataFrame(metric_rows).sort_values('Accuracy', ascending=False)
metrics_data
```

```python
# Confusion matrix for the best model
best_name = final_data.sort_values('Accuracy', ascending=False).iloc[0]['Models']
best_pred = predictions[best_name]

cm = confusion_matrix(y_test, best_pred)

plt.figure(figsize=(9, 7))
sns.heatmap(
    cm,
    annot=True,
    fmt='d',
    xticklabels=label_encoder.classes_,
    yticklabels=label_encoder.classes_
)
plt.xlabel('Predicted')
plt.ylabel('Actual')
plt.title(f'Confusion Matrix - {best_name}')
plt.show()

print("Best Model:", best_name)
```

## 14 - Save The Model

```python
import joblib
import os

os.makedirs('models', exist_ok=True)

best_model = {
    'Logistic Regression': lr,
    'SVM': svm,
    'Decision Tree': dt,
    'K-Nearest Neighbour': knn,
    'Random Forest': rf
}[best_name]

# Retrain the selected model on all available feature data before deployment.
best_model.fit(X, y)

joblib.dump(best_model, 'models/music_genre_model.pkl')
joblib.dump(label_encoder, 'models/label_encoder.pkl')
joblib.dump(feature_columns, 'models/feature_columns.pkl')

print("Saved:", best_name)
print("Model path: models/music_genre_model.pkl")
print("Encoder path: models/label_encoder.pkl")
print("Feature order path: models/feature_columns.pkl")
```
